**4. Exploratory Data Analysis**

Import the libraries and load the dataset

```python
#Import the libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('marketing_campaign.csv')
```

**Task 1 : General Data Overview**

a. Check the first few rows of the dataset to understand its structure

```python
print(df.head())
```

b.Check the data type of each column

```python
print(df.info())
```

c.Check for any missing values in the dataset

```python
print(df.isna().sum())
```

```python
# Handle missing values in Income column

median_income = df['Income'].median()
df['Income']=df['Income'].fillna(median_income)
```

```python
# Check for any other missing values
print(df.isnull().sum())
```

**Task 2 : Descriptive Statistics**

a. Compute summary statistics for numeric columns(mean, medium,min, max and standard deviation)

```python
print(df.describe())
```

b.Explore the distribution of numerical variables using histograms or box plots

```python
# Define numerical columns
num_cols = df.select_dtypes(include=np.number).columns

# Histograms for numerical variables
for col in num_cols:
    plt.figure(figsize=(8, 6))
    sns.histplot(df[col], kde=True, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Box plots for numerical variables
for col in num_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=df[col], color='lightgreen')
    plt.title(f'Box Plot of {col}')
    plt.xlabel(col)
    plt.show()
```

**Task 3 : Univariate Analysis**

a. Explore the distribution of each numerical variable using histograms or kernel density plots

```python
# Numerical variables
num_cols = df.select_dtypes(include=np.number).columns
for col in num_cols:
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

b.Explore the distribution of each categorical variable using bar plots or pie charts

```python
# Categorical variables
cat_cols = df.select_dtypes(exclude=np.number).columns
for col in cat_cols:
    sns.countplot(data=df, x=col)
    plt.title(col)
    plt.xticks(rotation=45)
    plt.show()
```

c.Identify outliers in numerical variables using box plots or scatter plots

```python
num_col = df.select_dtypes(include=np.number).columns

# Box plots for numerical variables
for col in num_col:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=df[col], color='pink')
    plt.title(f'Box Plot of {col}')
    plt.xlabel(col)
    plt.show()
```

**Task 4 : Bivariate Analysis**

a. Explore the relationship between numerical variables and the target variable(Response) using scatter plots or correlation matrices

```python
# Numerical variables
num_cols = df.select_dtypes(include=np.number).columns
# Numerical vs. Target variable
for col in num_cols:
    sns.scatterplot(data=df, x='Response', y=col)
    plt.title(col + ' vs. Response')
    plt.show()
```

a. Explore the relationship between categorical variables and the target variable(Response) using box plots or cchi-squared tests

```python
# Categorical variables
cat_cols = df.select_dtypes(exclude=np.number).columns
# Categorical vs. Target variable
for col in cat_cols:
    sns.countplot(data=df, x=col, hue='Response')
    plt.title(col + ' vs. Response')
    plt.xticks(rotation=45)
    plt.show()
```

c.Explore the relationship between numerical and categorical variables

```python
# Correlation Analysis
#convert category to numeric codes
df['Education'] = df['Education'].astype('category').cat.codes
df['Marital_Status'] = df['Marital_Status'].astype('category').cat.codes
df['Dt_Customer']  = df['Dt_Customer'].astype('category').cat.codes

corr_matrix = df.select_dtypes([np.number]).corr()
plt.figure(figsize=(16,16))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

```
